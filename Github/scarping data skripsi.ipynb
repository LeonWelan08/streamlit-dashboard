```python
import urllib.parse
import requests
import pandas as pd

# Daftar kata kunci
keywords = ['pemilu 2024']

# Loop untuk setiap kata kunci
for string in keywords:
    title, snippet, url, source, date = [], [], [], [], []

    # Loop untuk paginasi (setiap halaman memuat 100 berita)
    for i in [0, 100, 200, 300, 400, 500, 600, 700, 800, 900, 1000]:
        params = {
            'q': string,
            'tbm': "nws",
            'location': "Indonesia",
            'api_key': "9222eca576aca107dd76fe0258db698348393b0044af6e7da908bd58f5c4387b",  # Simpan di environment variable jika memungkinkan
            "start": i,
            'num': "1000",
        }

        # Encode parameter dan kirim permintaan HTTP
        param = urllib.parse.urlencode(params)
        response = requests.get('https://serpapi.com/search.json?' + param)
        data = response.json()

        # Ekstrak data dari respons
        for item in data.get('news_results', []):
            url.append(item.get('link'))
            title.append(item.get('title'))
            snippet.append(item.get('snippet'))
            source.append(item.get('source'))
            date.append(item.get('date'))

    # Buat DataFrame dari data yang diperoleh
    df = pd.DataFrame({
        'url': url,
        'judul': title,
        'konten': snippet,
        'sumber': source,
        'tanggal': date,
    })

    # Simpan DataFrame ke file Excel
    filename = f'Keyword_{string}.xlsx'
    df.to_excel(filename, index=False)

    # Opsi: Membaca kembali file Excel untuk verifikasi
    df_loaded = pd.read_excel(filename)
    print(df_loaded.head())  # Tampilkan beberapa baris pertama untuk memastikan
```
