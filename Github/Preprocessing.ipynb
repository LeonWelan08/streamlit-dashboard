```python
import pandas as pd
import re
import nltk
from nltk.tokenize import word_tokenize
from nltk.corpus import stopwords
from Sastrawi.Stemmer.StemmerFactory import StemmerFactory

# Download NLTK (sekali saja)
nltk.download('punkt')
nltk.download('punkt_tab')
nltk.download('stopwords')

# Load data
df = pd.read_excel("Keyword_pemilu 2024 (1).xlsx")

# Inisialisasi
stop_words = set(stopwords.words('indonesian'))
factory = StemmerFactory()
stemmer = factory.create_stemmer()

# Normalisasi kata tidak baku
normalisasi = {
    "gk": "tidak",
    "ga": "tidak",
    "nggak": "tidak",
    "bgt": "banget",
    "aja": "saja"
}

def preprocess(text):
    # Hindari error jika kosong
    text = str(text)
    
    # 1. Case Folding
    text = text.lower()
    
    # 2. Cleaning
    text = re.sub(r'[^a-zA-Z\s]', ' ', text)  # hapus angka & simbol
    
    # 3. Tokenization
    tokens = word_tokenize(text)
    
    # 6. Normalisasi
    tokens = [normalisasi.get(word, word) for word in tokens]
    
    # 4. Stopword Removal
    tokens = [word for word in tokens if word not in stop_words]
    
    # 5. Stemming
    tokens = [stemmer.stem(word) for word in tokens]
    
    return " ".join(tokens)

# Bersihkan kolom 'konten'
df['konten_bersih'] = df['konten'].apply(preprocess)

# Simpan hasil
df.to_excel("hasil_preprocessing_konten.xlsx", index=False)

print(df[['konten', 'konten_bersih']].head())
```

```python
import pandas as pd
import re
from nltk.corpus import stopwords
from Sastrawi.Stemmer.StemmerFactory import StemmerFactory

# Load data
df = pd.read_excel("Keyword_pemilu 2024 (1).xlsx")

# Inisialisasi
stop_words = set(stopwords.words('indonesian'))
factory = StemmerFactory()
stemmer = factory.create_stemmer()

# Normalisasi
normalisasi = {
    "gk": "tidak",
    "ga": "tidak",
    "nggak": "tidak",
    "bgt": "banget",
    "aja": "saja",
    "id" : "indonesia"
}

singkatan = {
    "kpu": "komisi pemilihan umum",
    "bawaslu": "badan pengawas pemilu",
    "dpr": "dewan perwakilan rakyat",
    "dprd": "dewan perwakilan rakyat daerah",
    "ri": "republik indonesia",
    "capres": "calon presiden",
    "cawapres": "calon wakil presiden",
    "pemilu": "pemilihan umum"
}

# Hindari null
df['konten'] = df['konten'].fillna('')

# =========================
# 1. Case Folding
# =========================
df_step1 = df.copy()
df_step1['case_folding'] = df_step1['konten'].str.lower()
df_step1.to_excel("step1_case_folding.xlsx", index=False)

# =========================
# 2. Cleaning
# =========================
df_step2 = df_step1.copy()
df_step2['cleaning'] = df_step2['case_folding'].apply(
    lambda x: re.sub(r'[^a-zA-Z\s]', ' ', x)
)
df_step2.to_excel("step2_cleaning.xlsx", index=False)

# =========================
# 3. Tokenization
# =========================
df_step3 = df_step2.copy()
df_step3['tokenization'] = df_step3['cleaning'].apply(lambda x: x.split())
df_step3.to_excel("step3_tokenization.xlsx", index=False)

# =========================
# 6. Normalisasi
# =========================
df_step4 = df_step3.copy()
df_step4['normalisasi'] = df_step4['tokenization'].apply(
    lambda tokens: [normalisasi.get(word, word) for word in tokens]
)
df_step4.to_excel("step4_normalisasi.xlsx", index=False)

# =========================
# 4. Stopword Removal
# =========================
df_step5 = df_step4.copy()
df_step5['stopword'] = df_step5['normalisasi'].apply(
    lambda tokens: [word for word in tokens if word not in stop_words]
)
df_step5.to_excel("step5_stopword.xlsx", index=False)

# =========================
# 5. Stemming
# =========================
df_step6 = df_step5.copy()
df_step6['stemming'] = df_step6['stopword'].apply(
    lambda tokens: [stemmer.stem(word) for word in tokens]
)
df_step6['hasil_preprocessing'] = df_step6['stemming'].apply(lambda x: " ".join(x))
df_step6.to_excel("step6_stemming_final.xlsx", index=False)

print("Semua file preprocessing berhasil disimpan!")
```

```python
import pandas as pd

df = pd.read_excel("step3_tokenization.xlsx")

# karena dari Excel bentuknya string → ubah ke list
import ast
df['tokenization'] = df['tokenization'].apply(ast.literal_eval)

# Normalisasi
normalisasi = {
    "gk": "tidak",
    "ga": "tidak",
    "nggak": "tidak",
    "bgt": "banget",
    "aja": "saja",
    "id" : "indonesia",
    "yg" : "yang"
}

singkatan = {
    "bawaslu": "badan pengawas pemilu",
    "dpr": "dewan perwakilan rakyat",
    "dprd": "dewan perwakilan rakyat daerah",
    "ri": "republik indonesia",
    "capres": "calon presiden",
    "cawapres": "calon wakil presiden",
    "pemilu": "pemilihan umum"
}


def normalisasi_kata(tokens):
    hasil = []
    for word in tokens:
        if word in singkatan:
            hasil.extend(singkatan[word].split())
        elif word in normalisasi:
            hasil.append(normalisasi[word])
        else:
            hasil.append(word)
    return hasil

df['normalisasi'] = df['tokenization'].apply(normalisasi_kata)

df.to_excel("step4_normalisasi.xlsx", index=False)
```

```python
import pandas as pd
import ast
from nltk.corpus import stopwords

df = pd.read_excel("step4_normalisasi.xlsx")

df['normalisasi'] = df['normalisasi'].apply(ast.literal_eval)

stop_words = set(stopwords.words('indonesian'))
# kata yang JANGAN dihapus
pengecualian = ['umum', 'hasil', 'data']

# hapus dari stopword
stop_words = stop_words - set(pengecualian)

df['stopword'] = df['normalisasi'].apply(
    lambda tokens: [word for word in tokens if word not in stop_words]
)

df.to_excel("step5_stopword.xlsx", index=False)
```
